### PyTorch里添加批量归一化

```python
import torch.nn as nn

class mymodel(nn.Module()):
    def __init__(self,input_dim):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(input_dim, 128), 
            nn.BatchNorm1d(128), 
            nn.ReLU(), 

            nn.Linear(128, 64), 
            nn.BatchNorm1d(64), 
            nn.ReLU(), 
            
            nn.Linear(64, 10),   
        )
```
